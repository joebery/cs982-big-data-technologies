```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv("train.csv")
```

```python
titanic.head(15)
```

```python
titanic.Age.plot(kind="box", color="red")


# titanic.Fare.plot(kind="box")
```

```python
titanic['Survived'].value_counts().plot.bar()
plt.ylabel("No. of passengers")
plt.title('How many passengers survived')
plt.show()
```

```python
titanic['Pclass'].value_counts().sort_values(ascending=False).plot.barh()
plt.title('How many passengers where in each class?')
plt.show()
```

```python
sns.set_style("whitegrid")
sns.kdeplot(np.array(titanic['SibSp']))
```

```python
titanic["Age"].plot(kind="hist")
plt.title("Histogram of age")
plt.ylabel("Number of people")
plt.show()
```

```python
sns.kdeplot(np.array(titanic['Age']))
```

```python
titanic['Survived'].value_counts().plot.bar(color = "red")
plt.ylabel('Count')
plt.show()
```

```python
titanic["Sex"].value_counts().plot.pie(autopct="%1.1f%%")
plt.axis('equal')
plt.title("male vs female")
plt.ylabel('just because')
plt.show()
```

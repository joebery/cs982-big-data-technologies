```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
pokemon = pd.read_csv("pokemon_alopez247.csv")

pokemon.groupby(['Type_1','Type_2']).size().unstack().plot(kind='bar',stacked=True)
ax = plt.gca()
ax.legend_ = None
plt.show()
```

```python
a = [67, 10, 30, 20, 25]
b = [30, 42, 11, 17, 5]
X = np.arange(5)
width = 0.35
fig, ax = plt.subplots()
ax.bar(X, a, width, color='blue')
ax.bar(X+width, b, width, color='y')
ax.set_xticks(X + width / 2)
ax.set_xticklabels(('Cat1', 'Cat2', 'Cat3', 'Cat4', 'Cat5'))
plt.show()
```

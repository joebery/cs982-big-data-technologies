```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL.Image import alpha_composite
```

```python
titanic = pd.read_csv("train.csv")
```

```python
titanic.head(15)
```

```python
titanic.groupby(['Pclass','Sex']).size().unstack().plot(kind='bar', stacked=True)
plt.ylabel('No who Survived')
plt.title("Relationship between survivors in each class and the Gender")
plt.show()
```

```python
corr = titanic.corr(numeric_only=True)
sns.heatmap(corr)

```

```python
plt.scatter(titanic["Age"], titanic["Pclass"] ,alpha=0.5)
plt.ylabel("class")
plt.xlabel("age")
plt.show()
```

```python
plt.scatter(titanic["Age"], titanic["SibSp"] ,alpha=0.5)
plt.ylabel("siblings")
plt.xlabel("age")
plt.show()
```

```python
sns.pairplot(titanic[["Age", "Fare"]])
```
